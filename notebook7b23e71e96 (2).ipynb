```python
import numpy as np

# 1. Pergatitja e mundesive per secilin stacion
candidates = {s: [] for s in range(1, udp.n_stations + 1)}

for triplet, values in udp.flat_db.items():

    asteroid_id, station_id, opportunity_id = triplet

    start, end = windows[station_id - 1]

    if start <= values[0] <= end:
        candidates[station_id].append(
            (asteroid_id, opportunity_id, values)
        )


# 2. Inicializimi i materialeve
masses = np.zeros((udp.n_stations, 3))

used_asteroids = set()
assignments = {}

# 3. Algoritmi Greedy
while True:

    available_stations = []

    for station_id in range(1, udp.n_stations + 1):

        available = [
            candidate
            for candidate in candidates[station_id]
            if candidate[0] not in used_asteroids
        ]

        if available:
            station_min = min(masses[station_id - 1])

            available_stations.append(
                (station_min, station_id, available)
            )

    if not available_stations:
        break

    # Zgjedhim stacionin me materialin minimal me te ulet
    _, station_id, available = min(
        available_stations,
        key=lambda x: (x[0], sum(masses[x[1] - 1]))
    )

    # Zgjedhim asteroidin qe e permireson balancen
    best = max(
        available,
        key=lambda x: (
            min(masses[station_id - 1] + x[2][1:]),
            sum(x[2][1:])
        )
    )

    asteroid_id, opportunity_id, values = best

    assignments[asteroid_id] = (
        station_id,
        opportunity_id + 1
    )

    used_asteroids.add(asteroid_id)

    masses[station_id - 1] += values[1:]


# 4. Ndertimi i kromozomit
window_values = [
    value
    for pair in windows
    for value in pair
]

schedule = []

for asteroid_id, (station_id, opportunity_id) in assignments.items():
    schedule.extend([
        asteroid_id,
        station_id,
        opportunity_id
    ])

chromosome = udp.convert_to_chromosome(
    np.array(window_values + schedule)
)


# 5. Vleresimi zyrtar
result = udp.fitness(chromosome)

print("===== REZULTATI I ALGORITMIT GREEDY =====")
print("Asteroide te caktuar:", len(assignments))
print("Asteroide te pacaktuar:", 340 - len(assignments))

print("\nFitness:", result[0])
print("Materiali minimal:", -result[0])

print("\nKufizimet:")
print("Kufizimi 1:", result[1])
print("Kufizimi 2:", result[2])
print("Kufizimi 3:", result[3])
print("Kufizimi 4:", result[4])

print("\n===== PERMBLEDHJA ZYRTARE =====")
udp.pretty(chromosome)
```

```python
import random
import numpy as np

def randomized_greedy(windows, seed):
    random.seed(seed)

    candidates = {s: [] for s in range(1, udp.n_stations + 1)}

    for triplet, values in udp.flat_db.items():
        asteroid_id, station_id, opportunity_id = triplet
        start, end = windows[station_id - 1]

        if start <= values[0] <= end:
            candidates[station_id].append(
                (asteroid_id, opportunity_id, values)
            )

    masses = np.zeros((udp.n_stations, 3))
    used = set()
    assignments_new = {}

    while True:
        available_stations = []

        for station_id in range(1, udp.n_stations + 1):
            available = [
                c for c in candidates[station_id]
                if c[0] not in used
            ]

            if available:
                available_stations.append((station_id, available))

        if not available_stations:
            break

        station_id, available = min(
            available_stations,
            key=lambda x: min(masses[x[0] - 1])
        )

        random.shuffle(available)

        best = max(
            available,
            key=lambda x: min(
                masses[station_id - 1] + x[2][1:]
            )
        )

        asteroid_id, opportunity_id, values = best

        assignments_new[asteroid_id] = (
            station_id, opportunity_id + 1
        )

        used.add(asteroid_id)
        masses[station_id - 1] += values[1:]

    window_values = [v for pair in windows for v in pair]
    schedule = []

    for asteroid_id, (station_id, opportunity_id) in assignments_new.items():
        schedule.extend([asteroid_id, station_id, opportunity_id])

    chromosome_new = udp.convert_to_chromosome(
        np.array(window_values + schedule)
    )

    return chromosome_new, assignments_new
```
