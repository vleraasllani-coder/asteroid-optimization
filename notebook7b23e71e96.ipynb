```python
import requests
import os

# Linku i databazes
url = "https://api.optimize.esa.int/data/spoc/scheduling/candidates.txt"

# Krijimi i folderit data
os.makedirs("data", exist_ok=True)

# Shkarkimi i databazes
response = requests.get(url)
response.raise_for_status()

# Ruajtja e databazes
with open("data/candidates.txt", "w", encoding="utf-8") as file:
    file.write(response.text)

print("Databaza u shkarkua me sukses!")

print("Databaza u shkarkua me sukses!")
print("Madhësia e databazës:", len(response.text), "karaktere")

import ast

# Leximi i databazes
data = ast.literal_eval(response.text)

# Numri i asteroideve
print("Numri i asteroideve:", len(data))

# Numri i stacioneve per asteroidin e pare
print("Stacionet per asteroidin e pare:", len(data["1"]))

# Numrimi i mundesive te transferimit
total = 0

for asteroid in data.values():
    for station in asteroid.values():
        total += len(station)

print("Numri total i mundesive te transferimit:", total)

# Analiza e mundesive per asteroid
print("\nMundesite per 5 asteroidet e pare:")

for asteroid_id, stations in list(data.items())[:5]:
    total_mundesi = 0

    for trips in stations.values():
        total_mundesi += len(trips)

    print("Asteroidi", asteroid_id, "ka", total_mundesi, "mundesi transferimi")
    print("\nMundësitë e transferimit për asteroidin 1, stacionin 1:")

print(data["1"]["9"])
```
